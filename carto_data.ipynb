## import des fichiers avec pandas

```python
import pandas as pd

df_erp=pd.read_excel("data/Fichier_erp.xlsx")
df_cms=pd.read_excel("data/Fichier_web.xlsx")
df_liaison=pd.read_excel("data/fichier_liaison.xlsx")

```

```python
print("Types ERP :")
print(df_erp.dtypes)

print("\nTypes CMS :")
print(df_cms.dtypes)

print("\nTypes liaison :")
print(df_liaison.dtypes)
```

# Synthese de la carto

## ERP

- Volumétrie : **825 lignes et 5 colonnes**.
- Clé probable : `product_id`.
- `product_id` contient **825 valeurs uniques pour 825 lignes**.
- Aucune valeur manquante n'est présente dans le fichier.
- Aucun doublon complet n'a été détecté.
- Aucun doublon n'a été détecté sur `product_id`.
- `product_id` peut donc être considéré comme une **clé candidate**.
- Les données contiennent notamment le prix, le stock et le statut du stock.

## CMS / Web

- Volumétrie : **1 513 lignes et 28 colonnes**.
- Clé probable : `sku`.
- `sku` contient **714 valeurs uniques**.
- **85 valeurs de `sku` sont manquantes**.
- Après exclusion des valeurs manquantes, **714 doublons sont présents sur `sku`**.
- **82 doublons complets** sont également présents dans le fichier.
- Plusieurs autres colonnes contiennent des valeurs manquantes.
- La colonne `sku` n'est donc pas unique dans le fichier brut et devra être nettoyée avant la jointure.

## Fichier de liaison

- Volumétrie : **825 lignes et 2 colonnes**.
- Colonnes de liaison : `product_id` et `id_web`.
- `product_id` contient **825 valeurs uniques**, sans valeur manquante ni doublon.
- `id_web` contient **734 valeurs uniques** et **91 valeurs manquantes**.
- Parmi les `id_web` renseignés, aucun doublon n'a été détecté.
- Aucun doublon complet n'a été détecté dans le fichier.

## Contraintes identifiées

- Les trois sources sont fournies au format Excel `.xlsx`.
- L'ERP utilise `product_id` comme identifiant.
- Le CMS utilise `sku` comme identifiant.
- Le fichier de liaison permet de relier `product_id` à `id_web`.
- `product_id` est de type entier dans l'ERP et dans le fichier de liaison.
- `sku` et `id_web` sont de type texte / `object`.
- Le principal problème de qualité concerne le CMS : valeurs manquantes et répétitions de `sku`.
- Le fichier de liaison contient également des valeurs manquantes dans `id_web`.
- Ces problèmes devront être traités avant les jointures afin d'éviter des données manquantes ou dupliquées.

## Conclusion

L'ERP présente une structure propre avec un `product_id` unique et complet.
Le CMS nécessite un nettoyage avant utilisation, notamment à cause des `sku` manquants et non uniques.

Le fichier de liaison permet de faire le lien entre les identifiants ERP et Web, mais certaines valeurs `id_web` sont absentes.
